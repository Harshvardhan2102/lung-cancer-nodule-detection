```python
from ultralytics import YOLO

# Load pretrained model
model = YOLO('yolov8n.pt')

# Train with GPU + more dataloader workers
results = model.train(
    data='/kaggle/working/data.yml',
    epochs=20,
    device=0,        # use GPU (0 = first GPU)
    workers=4,       # increase workers (try 2 or 4, Kaggle supports it)
    batch=16         # optional: set batch size
)
```
